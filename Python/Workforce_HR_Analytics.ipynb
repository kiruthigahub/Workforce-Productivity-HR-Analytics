#Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

# Load the datasets

```python
employees = pd.read_csv("Employees_cleaned.csv")
departments = pd.read_csv("Departments_cleaned.csv")
attendance = pd.read_csv("Attendance_cleaned.csv")
performance = pd.read_csv("Performance_cleaned.csv")

print("All datasets loaded successfully!")
```

#Check dataset dimensions

```python
print("Employees:", employees.shape)
print("Departments:", departments.shape)
print("Attendance:", attendance.shape)
print("Performance:", performance.shape)
```

Question 29 – Employee Count by Department (Bar Chart)

```python
employees["Department"].value_counts().plot(kind="bar")

plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Question 30 – Attendance Status Distribution (Pie Chart)
